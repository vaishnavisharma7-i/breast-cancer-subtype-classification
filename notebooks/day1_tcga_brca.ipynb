```python
import pandas as pd

expression = pd.read_csv(
    "../data/raw/BRCA.exp.547.med.txt",
    sep="\t"
)

pam50 = pd.read_csv(
    "../data/raw/BRCA.547.PAM50.SigClust.Subtypes.txt",
    sep="\t"
)

print("Expression shape:", expression.shape)
print("PAM50 shape:", pam50.shape)

print(expression.head())
print(pam50.head())
```

```python
print(pam50.columns.tolist())
print(pam50.head(10).to_string())
print(pam50.iloc[:, -1].value_counts())
```

```python
# combining two biological dataset 
expression_samples = set(expression.columns[1:])
pam50_samples = set(pam50["Sample"])

print("Expression samples:", len(expression_samples))
print("PAM50 samples:", len(pam50_samples))
```

```python
common_samples = expression_samples & pam50_samples

print("Common samples:", len(common_samples))
```

```python
print("Expression only:", len(expression_samples - pam50_samples))
print("PAM50 only:", len(pam50_samples - expression_samples))
```
