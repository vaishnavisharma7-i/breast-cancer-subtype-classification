```python
import pandas as pd

phenotype = pd.read_csv(
    "../data/raw/brca_pam50",
    sep="\t"
)

print(phenotype.shape)
print(phenotype.head())
print(phenotype.columns)
```

```python
print(phenotype["PAM50"].value_counts())
```

```python
phenotype["sample"].duplicated().sum()

phenotype["sample"].value_counts().head(20)


duplicates = phenotype[
    phenotype["sample"].duplicated(keep=False)
]

print(duplicates)
```

```python
print(phenotype[phenotype["PAM50"] == "Unknown"])
print(phenotype[phenotype["PAM50"] == "Normal"])

```

```python
expression_samples = expression.columns[1:]
phenotype_samples = phenotype["sample"].unique()

print("Expression samples:", len(expression_samples))
print("Unique phenotype samples:", len(phenotype_samples))
```

```python
common_samples = set(expression_samples) & set(phenotype_samples)
print("Common samples:", len(common_samples))
```

```python
phenotype_clean = phenotype.drop_duplicates(subset="sample")

print("Original rows:", len(phenotype))
print("Unique samples:", len(phenotype_clean))
```

```python
duplicate_check = (
    phenotype.groupby("sample")["PAM50"]
    .nunique()
)

print(duplicate_check.value_counts())

print(phenotype_clean["PAM50"].value_counts())
```
